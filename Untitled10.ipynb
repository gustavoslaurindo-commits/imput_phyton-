```python
### verficacoes

def verificar_aprovacao(nota):
    if nota >= 6:
        return " você foi Aprovado"
    else:
        return " Você foi Reprovado"

print(verificar_aprovacao(7))
print(verificar_aprovacao(4))
```

```python
## Realizando a soma

def soma_lista(numeros):
    soma = 0
    for numero in numeros:
        soma = soma + numero
    return soma

lista_numeros = [10, 20, 30, 40]
print(soma_lista(lista_numeros))
```

```python


def calcula_media(notas):

    soma = 0
    for nota in notas:
        soma = soma + nota

    return soma / len(notas)

notas_A = [7, 8, 6]
notas_B = [5, 9, 5]

print(f"Média A: {calcula_media(notas_A):.2f}")
print(f"Média B: {calcula_media(notas_B):.2f}")
```
