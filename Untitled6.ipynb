```python
##   append

NOMES = ["Ana", "Carlos"]
NOMES.append("João")
print (NOMES)
```

```python
##   remove

NOMES = ["Ana", "Carlos"]
NOMES.remove("Carlos")
print (NOMES)
```

```python
## pop
NOMES = ["Ana", "Carlos", "João"]
NOMES.pop(2)
print(NOMES)
```

```python
## lenght === len  função

NOMES = ["Ana", "Carlos", "João"]
print(len(NOMES))
```

```python
## sort

NOMES = ["João", "Ana", "Daniel", "Anderson"]
NOMES.sort()
print(NOMES)

números = [3,5,1,2,6,4]
números.sort()
print(números)
```

```python
###### function reverse
NOMES = ["João", "Ana", "Daniel", "Anderson"]
NOMES.reverse()
print(NOMES)


```
